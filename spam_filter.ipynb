Train and evaluate a text classifier that labels SMS messages as `spam` or `ham` (not spam). This notebook downloads the public UCI dataset at runtime; the dataset is not committed to this repository.

## 1. Import libraries


```python
from pathlib import Path
import urllib.request
import zipfile
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, precision_score, recall_score, confusion_matrix, classification_report
```

## 2. Download and load the dataset

The UCI archive is downloaded to the runtime's temporary working directory.


```python
from pathlib import Path
import urllib.request
import zipfile
import pandas as pd

DATA_DIR = Path("data")
DATA_DIR.mkdir(exist_ok=True)

DATA_FILE = DATA_DIR / "SMSSpamCollection"
ZIP_FILE = DATA_DIR / "sms_spam_collection.zip"

# Download the dataset only if it is not already available.
if not DATA_FILE.exists():
    url = "https://archive.ics.uci.edu/static/public/228/sms+spam+collection.zip"
    urllib.request.urlretrieve(url, ZIP_FILE)

    with zipfile.ZipFile(ZIP_FILE, "r") as archive:
        archive.extract("SMSSpamCollection", DATA_DIR)

df = pd.read_csv(
    DATA_FILE,
    sep="\t",
    header=None,
    names=["label", "message"],
    encoding="utf-8"
)

print("Dataset shape:", df.shape)
display(df.head())
print("Missing values:\n", df.isna().sum())
```

## 3. Split the data

Use a stratified split so the spam/ham proportions are approximately preserved in both sets.


```python
X_train, X_test, y_train, y_test = train_test_split(
    df["message"],
    df["label"],
    test_size=0.2,
    random_state=42,
    stratify=df["label"]
)
print(f"Training messages: {len(X_train)}")
print(f"Testing messages: {len(X_test)}")
```

## 4. Vectorize text and train the model

Fit TF-IDF only on the training messages to avoid leaking test-set information into training.


```python
vec = TfidfVectorizer()
X_train_vec = vec.fit_transform(X_train)

model = MultinomialNB()
model.fit(X_train_vec, y_train)
```

## 5. Evaluate on held-out messages


```python
X_test_vec = vec.transform(X_test)
y_pred = model.predict(X_test_vec)

print(f"Accuracy:  {accuracy_score(y_test, y_pred):.4f}")
print(f"Precision: {precision_score(y_test, y_pred, pos_label='spam'):.4f}")
print(f"Recall:    {recall_score(y_test, y_pred, pos_label='spam'):.4f}")
print("\nClassification report:\n")
print(classification_report(y_test, y_pred, labels=["ham", "spam"], zero_division=0))
print("Confusion matrix (rows = actual, columns = predicted; order ham, spam):")
display(pd.DataFrame(confusion_matrix(y_test, y_pred, labels=["ham", "spam"]),
                     index=["actual ham", "actual spam"],
                     columns=["predicted ham", "predicted spam"]))
```

## 6. Try five challenging messages

```python
tests = [
    "Cl@im your fr33 g1ft card now",
    "Not free, just confirming lunch at 1pm",
    "Congratulations! You have won a free prize. Claim now!",
    "Your account is blocked. Verify your details immediately.",
    "Hi, could you please send me the meeting notes?"
]

test_predictions = model.predict(vec.transform(tests))
results = pd.DataFrame({
    "My message": tests,
    "Model said": test_predictions,
    "Should be": ["spam", "ham", "spam", "spam", "ham"],
    "Why it failed": ["", "", "", "", ""]
})
display(results)
```
